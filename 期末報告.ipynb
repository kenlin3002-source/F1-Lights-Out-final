```
# 1. 自動安裝 Gradio 與 Matplotlib
!pip install -q gradio matplotlib pandas numpy

import gradio as gr
import matplotlib.pyplot as plt
import numpy as np

# 【修正關鍵】使用絕對安全、不衝突的圖表樣式，並設定好字體大小
plt.style.use('default')
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['font.size'] = 10

# ==================== 【後端 AI 核心演算法與大腦】 ====================

# 功能一：核心課表與 AI 診斷生成
def triathlon_ai_coach(sport, fatigue, vo2max, total_time, intensity_zone):
    # --- A. AI 體能與疲勞度診斷邏輯 ---
    if fatigue >= 8:
        ai_analysis = (
            f"🚨 【AI 教練嚴重警告】\n"
            f"偵測到您當前的疲勞度高達 {fatigue}/10！此狀態下強行進行強度訓練，極易導致肌肉拉傷或過度訓練症候群。\n"
            f"💡 【建議】今日應嚴格執行「動態恢復」，不建議進行 Zone 3 以上的訓練。"
        )
    else:
        if vo2max >= 55:
            athlete_level = "精英級鐵人 (Elite)"
        elif 45 <= vo2max < 55:
            athlete_level = "進階運動員 (Advanced)"
        else:
            athlete_level = "健康規律運動者 (Fitness)"

        ai_analysis = (
            f"📈 【AI 體能動態診斷報告】\n"
            f"● 您的目前最大攝氧量 (VO2 Max) 為 {vo2max}，屬於：{athlete_level}。\n"
            f"● 當前疲勞度控制良好 ({fatigue}/10)，身體儲備能量充足，適合執行本次設定的 {intensity_zone} 訓練。\n"
            f"● 評估：本次 {total_time} 分鐘的訓練刺激，有助於進一步優化您的有氧底子與耐乳酸能力！"
        )

    # --- B. 智慧區塊化課表生成邏輯 (分段計算時間) ---
    if fatigue >= 8:
        warmup_t = int(total_time * 0.3)
        main_t = int(total_time * 0.4)
        cooldown_t = total_time - warmup_t - main_t

        warmup_detail = f"⏱️ {warmup_t} 分鐘\n● 極輕度輕鬆活動（慢速踩踏或散步）\n● 讓心率維持在 Zone 1 活性化範圍"
        main_detail = f"⏱️ {main_t} 分鐘\n● 靜態伸展與關節鬆動\n● 滾筒全身肌群放遜（特別加強下肢）"
        cooldown_detail = f"⏱️ {cooldown_t} 分鐘\n● 深呼吸冥想、多補充水分與電解質"
    else:
        warmup_t = int(total_time * 0.2)
        main_t = int(total_time * 0.65)
        cooldown_t = total_time - warmup_t - main_t

        if "跑步" in sport:
            warmup_detail = f"⏱️ {warmup_t} 分鐘\n● 輕鬆關節活動 + 慢跑\n● 漸進將心率提升至 Zone 1 頂端"
            main_detail = f"⏱️ {main_t} 分鐘\n● 核心主訓練：於【{intensity_zone}】維持穩定配速\n● 注意跑姿（步頻維持 180spm）"
            cooldown_detail = f"⏱️ {cooldown_t} 分鐘\n● 慢走放鬆 + 靜態拉伸\n● 降低心率並排除肌肉乳酸"
        elif "騎車" in sport:
            warmup_detail = f"⏱️ {warmup_t} 分鐘\n● 輕齒比、高迴轉速 (90 RPM) 踩踏\n● 喚醒大腿肌群血液循環"
            main_detail = f"⏱️ {main_t} 分鐘\n● 主訓練：在【{intensity_zone}】進行巡航\n● 保持上半身放鬆，專注於畫圓踩踏"
            cooldown_detail = f"⏱️ {cooldown_t} 分鐘\n● 降至最輕齒比輕鬆踩踏\n● 緩和心率與肌肉緊繃感"
        else:
            warmup_detail = f"⏱️ {warmup_t} 分鐘\n● 輕鬆任意式游 warm-up\n● 專注於水感與身體延伸"
            main_detail = f"⏱️ {main_t} 分鐘\n● 主訓練：於【{intensity_zone}】執行包夾練習或間歇\n● 注意划手動作與換氣節奏"
            cooldown_detail = f"⏱️ {cooldown_t} 分鐘\n● 100公尺輕鬆仰式或蛙式\n● 伸展肩關節與背闊肌"

    return ai_analysis, warmup_detail, main_detail, cooldown_detail

# 功能二：核心體能數據圖表生成
def generate_trend_chart(vo2max):
    # 建立畫布與圖表
    fig, ax = plt.subplots(figsize=(6, 3))
    weeks = ['Week 1', 'Week 2', 'Week 3', 'Week 4']

    # 用輸入的 VO2 Max 當作最新一週，往前推算歷史
    vo2_history = [float(vo2max) - 3, float(vo2max) - 1.5, float(vo2max) - 0.5, float(vo2max)]

    # 繪製高質感折線圖
    ax.plot(weeks, vo2_history, marker='o', color='#3498db', linewidth=3, markersize=8, label="VO2 Max")
    ax.fill_between(weeks, vo2_history, alpha=0.1, color='#3498db')

    ax.set_title("VO2 Max Monthly Progress Analysis", fontsize=11, fontweight='bold', pad=8)
    ax.set_ylabel("VO2 Max (ml/kg/min)", fontsize=9)
    ax.set_ylim(min(vo2_history) - 3, max(vo2_history) + 3)
    ax.grid(True, linestyle='--', alpha=0.5)

    # 調整間距避免文字被切到
    plt.tight_layout()
    return fig

# 整合一體化大腦：點擊按鈕或啟動時，一次把文字跟圖表全部回傳
def final_master_process(sport, fatigue, vo2max, total_time, intensity_zone):
    ai_analysis, warmup_detail, main_detail, cooldown_detail = triathlon_ai_coach(sport, fatigue, vo2max, total_time, intensity_zone)
    chart_fig = generate_trend_chart(vo2max)
    return ai_analysis, warmup_detail, main_detail, cooldown_detail, chart_fig


# ==================== 【Gradio UI 視覺化積木排版】 ====================

custom_css = """
.workout-card { background-color: #f8f9fa !important; border-left: 5px solid #3498db !important; border-radius: 6px !important; padding: 12px !important; margin-bottom: 8px !important; }
.ai-box { background-color: #ebf5fb !important; border: 1px solid #aec6cf !important; border-radius: 8px !important; }
"""

# 使用 Gradio Blocks 刻出與網頁設計圖一模一樣的排版
with gr.Blocks(theme=gr.themes.Soft(), css=custom_css) as demo:

    gr.Markdown("# 🏆 TriHub AI 智慧鐵人三項課表與體能分析系統")
    gr.Markdown("#### 運算思維期末報告展示原型 —— 輸入今日體能指標，由 Python 後端即時推算區塊化課表與圖表分析。")

    with gr.Row():

        # --- 左側欄位：使用者參數輸入區 ---
        with gr.Column(scale=1):
            gr.Markdown("### 🛠️ 第一步：請輸入今日數據")

            inp_sport = gr.Radio(
                ["跑步 (Running)", "騎車 (Cycling)", "游泳 (Swimming)"],
                label="🏃‍♂️ 選擇運動項目",
                value="跑步 (Running)"
            )

            inp_fatigue = gr.Slider(
                minimum=1, maximum=10, step=1,
                label="🧘‍♂️ 今日身體疲勞度自評 (1-10分)",
                value=3
            )

            inp_vo2max = gr.Slider(
                minimum=30, maximum=80, step=1,
                label="🫁 目前最大攝氧量 (VO2 Max)",
                value=48
            )

            inp_time = gr.Slider(
                minimum=20, maximum=180, step=5,
                label="⏱️ 預計總訓練時間 (分鐘)",
                value=60
            )

            inp_zone = gr.Dropdown(
                ["Zone 1 (輕鬆恢復區)", "Zone 2 (有氧耐力區)", "Zone 3 (節奏心率區)", "Zone 4 (乳酸閾值區)", "Zone 5 (無氧間歇區)"],
                label="🎯 預期目標訓練強度區間",
                value="Zone 2 (有氧耐力區)"
            )

            btn_submit = gr.Button("🚀 一鍵分析並生成課表", variant="primary")

        # --- 右側欄位：AI 分析與分段課表顯示區 ---
        with gr.Column(scale=1.5):
            gr.Markdown("### 📊 第二步：AI 教練診斷與課表輸出")

            # AI 診斷大腦
            out_ai = gr.Textbox(label="🧠 AI 體能與疲勞診斷大腦", lines=5, elem_classes="ai-box")

            # 課表區塊化顯示
            gr.Markdown("#### 📋 智慧配時分段訓練計畫")
            with gr.Row():
                out_warmup = gr.Textbox(label="🔥 1. 暖身階段 (Warm-up)", lines=5, elem_classes="workout-card")
                out_main = gr.Textbox(label="⚡ 2. 主訓練階段 (Main Set)", lines=5, elem_classes="workout-card")
                out_cooldown = gr.Textbox(label="🧊 3. 緩和階段 (Cool Down)", lines=5, elem_classes="workout-card")

            # 體能圖表趨勢
            gr.Markdown("#### 📈 個人體能數據視覺化趨勢")
            out_chart = gr.Plot(label="VO2 Max 歷史趨勢圖")

    # ==================== 【連結前後端：動作與初始值載入】 ====================

    # 1. 綁定點擊按鈕時的反應
    btn_submit.click(
        fn=final_master_process,
        inputs=[inp_sport, inp_fatigue, inp_vo2max, inp_time, inp_zone],
        outputs=[out_ai, out_warmup, out_main, out_cooldown, out_chart]
    )

    # 【優化重點】讓網頁一開啟，不需要點按鈕，就自動帶入預設值載入初始課表與圖表！
    demo.load(
        fn=final_master_process,
        inputs=[inp_sport, inp_fatigue, inp_vo2max, inp_time, inp_zone],
        outputs=[out_ai, out_warmup, out_main, out_cooldown, out_chart]
    )

# 啟動 Gradio 網頁介面
demo.launch(share=True)
```

```
# 1. 自動安裝 Gradio 與 Matplotlib
!pip install -q gradio matplotlib pandas numpy

import gradio as gr
import matplotlib.pyplot as plt
import numpy as np

# 使用安全乾淨的圖表基礎樣式
plt.style.use('default')
plt.rcParams['font.family'] = 'DejaVu Sans'

# ==================== 【後端智慧配速與課表大腦】 ====================

def generate_training_schedule(sport, run_type, fatigue, total_time):
    # 基礎配時比例：暖身 20%、主訓練 65%、緩和 15%
    warmup_t = int(total_time * 0.2)
    main_t = int(total_time * 0.65)
    cooldown_t = total_time - warmup_t - main_t

    # --- 1. AI 體能與疲勞度獨立診斷報告 ---
    if fatigue >= 8:
        ai_analysis = (
            f"🚨 【AI 教練疲勞警示】\n"
            f"當前身體疲勞度處於高點 ({fatigue}/10)！系統已自動啟動安全防護機制。\n"
            f"💡 【決策建議】今日全面調降訓練強度，強制執行「動態恢復課表」，避免肌肉拉傷與過度疲勞。"
        )
        # 疲勞度過高時的恢復配速/強度設定
        run_paces = {"warmup": "7:00", "main": "6:30", "cooldown": "7:30"}
        bike_stats = {"warmup": "90W (18 km/h)", "main": "110W (22 km/h)", "cooldown": "80W (15 km/h)"}
        swim_paces = {"warmup": "2:30", "main": "2:20", "cooldown": "2:45"}
    else:
        ai_analysis = (
            f"📈 【AI 體能動態診斷報告】\n"
            f"● 當前身體疲勞度控制良好 ({fatigue}/10)，能量儲備充足。\n"
            f"● 目標運動項目：{sport} " + (f"({run_type})" if sport == "跑步 (Running)" else "") + f"\n"
            f"● 評估：今日適合進行規律的結構化刺激，各階段專屬配速與目標強度已計算完成，請按表操課！"
        )
        # 正常狀態下的配速指標 (根據不同運動與類型調整)
        if "跑步" in sport:
            if run_type == "有氧跑":
                run_paces = {"warmup": "6:30", "main": "5:45", "cooldown": "6:45"}
            elif run_type == "節奏跑":
                run_paces = {"warmup": "6:15", "main": "5:00", "cooldown": "6:30"}
            elif run_type == "漸速跑":
                run_paces = {"warmup": "6:30", "main": "5:15", "cooldown": "6:15"}
            elif run_type == "變速跑":
                run_paces = {"warmup": "6:20", "main": "5:30", "cooldown": "6:30"}
            else: # 間歇跑
                run_paces = {"warmup": "6:15", "main": "4:30", "cooldown": "6:30"}

        bike_stats = {"warmup": "120W (22 km/h)", "main": "180W (30 km/h)", "cooldown": "100W (20 km/h)"}
        swim_paces = {"warmup": "2:15", "main": "1:50", "cooldown": "2:25"}

    # --- 2. 根據運動項目組合區塊化文字與單位 ---
    if "跑步" in sport:
        y_label = "Pace (min/km)"
        y_data = [float(run_paces["warmup"].replace(':', '.')), float(run_paces["main"].replace(':', '.')), float(run_paces["cooldown"].replace(':', '.'))]

        warmup_detail = f"⏱️ {warmup_t} 分鐘\n● 強度：輕鬆活動\n● 目標配速：{run_paces['warmup']} 分/公里\n● 穩定喚醒下肢肌群"
        main_detail = f"⏱️ {main_t} 分鐘\n● 強度：核心主訓練 ({run_type})\n● 目標配速：{run_paces['main']} 分/公里\n● 注意穩定呼吸與步頻"
        cooldown_detail = f"⏱️ {cooldown_t} 分鐘\n● 強度：慢走緩和\n● 目標配速：{run_paces['cooldown']} 分/公里\n● 協助肌肉排除乳酸"

    elif "騎車" in sport:
        y_label = "Power (Watts)"
        y_data = [int(bike_stats["warmup"].split('W')[0]), int(bike_stats["main"].split('W')[0]), int(bike_stats["cooldown"].split('W')[0])]

        warmup_detail = f"⏱️ {warmup_t} 分鐘\n● 強度：高迴轉速熱身\n● 目標指標：{bike_stats['warmup']}\n● 建立流暢踩踏迴轉"
        main_detail = f"⏱️ {main_t} 分鐘\n● 強度：巡航耐力\n● 目標指標：{bike_stats['main']}\n● 保持上半身放鬆穩定輸出"
        cooldown_detail = f"⏱️ {cooldown_t} 分鐘\n● 強度：輕齒比緩和\n● 目標指標：{bike_stats['cooldown']}\n● 降低雙腿肌肉緊繃感"

    else: # 游泳
        y_label = "Pace (min/100m)"
        y_data = [float(swim_paces["warmup"].replace(':', '.')), float(swim_paces["main"].replace(':', '.')), float(swim_paces["cooldown"].replace(':', '.'))]

        warmup_detail = f"⏱️ {warmup_t} 分鐘\n● 強度：水感熱身\n● 目標配速：{swim_paces['warmup']} /100m\n● 專注身體延伸與滑行"
        main_detail = f"⏱️ {main_t} 分鐘\n● 強度：有氧主段\n ---\n● 目標配速：{swim_paces['main']} /100m\n● 維持划手節奏與換氣穩定"
        cooldown_detail = f"⏱️ {cooldown_t} 分鐘\n● 強度：輕鬆游緩和\n● 目標配速：{swim_paces['cooldown']} /100m\n● 放鬆肩關節與背部肌群"

    # --- 3. 繪製精美動態強度/配速圖表 ---
    fig, ax = plt.subplots(figsize=(7, 3))
    phases = ['Warm-up', 'Main Set', 'Cool Down']

    # 繪製圖表步進樣式，展現課表段落感
    ax.plot(phases, y_data, marker='o', color='#e74c3c' if "騎車" in sport else '#3498db', linewidth=3, label="Target Intensity")
    ax.fill_between(phases, y_data, alpha=0.15, color='#e74c3c' if "騎車" in sport else '#3498db')

    ax.set_title(f"{sport} - Workout Intensity Profile", fontsize=12, fontweight='bold', pad=10)
    ax.set_ylabel(y_label, fontsize=10)
    ax.grid(True, linestyle='--', alpha=0.5)

    # 針對跑步/游泳配速反轉 Y 軸（因為數字越小代表速度越快）
    if "跑步" in sport or "游泳" in sport:
        ax.invert_yaxis()

    plt.tight_layout()

    return ai_analysis, warmup_detail, main_detail, cooldown_detail, fig

# 控制當運動項目改變時，是否顯示跑步專屬下拉選單
def update_run_type_visibility(sport):
    if sport == "跑步 (Running)":
        return gr.update(visible=True)
    return gr.update(visible=False)


# ==================== 【Gradio UI 介面視覺排版】 ====================

custom_css = """
.workout-card { background-color: #fdfefe !important; border-top: 4px solid #3498db !important; border-radius: 8px !important; padding: 12px !important; }
.ai-box { background-color: #f4f6f7 !important; border: 1px solid #d5dbdb !important; border-radius: 8px !important; font-size: 14px !important; }
"""

with gr.Blocks(theme=gr.themes.Soft(), css=custom_css) as demo:

    gr.Markdown("# 📋 訓練課表產生器")
    gr.Markdown("#### 運算思維期末專題原型展示 —— 由上往下的介面驅動開發，即時運算多運動項目之智慧配時與指標分析。")

    # 【第一區塊：參數輸入欄】排版寬敞不擁擠
    with gr.Row():
        with gr.Column():
            gr.Markdown("### 🛠️ 設定今日運動指標")
            inp_sport = gr.Radio(
                ["跑步 (Running)", "騎車 (Cycling)", "游泳 (Swimming)"],
                label="🏃‍♂️ 選擇運動項目",
                value="跑步 (Running)"
            )

            # 跑步專屬類型選單
            inp_run_type = gr.Dropdown(
                ["有氧跑", "節奏跑", "漸速跑", "變速跑", "間歇跑"],
                label="🎯 跑步訓練類型選擇",
                value="有氧跑",
                visible=True
            )

            inp_fatigue = gr.Slider(
                minimum=1, maximum=10, step=1,
                label="🧘‍♂️ 今日身體疲勞度自評 (1-10分)",
                value=3
            )

            inp_time = gr.Slider(
                minimum=20, maximum=180, step=5,
                label="⏱️ 預計總訓練時間 (分鐘)",
                value=60
            )

            btn_submit = gr.Button("🚀 一鍵分析並生成課表", variant="primary")

    # 連動邏輯：選跑步才看得到五大類型下拉選單
    inp_sport.change(fn=update_run_type_visibility, inputs=inp_sport, outputs=inp_run_type)

    # 【第二區塊：AI 分析獨立站一格（寬敞整行）】
    gr.Markdown("### 📊 AI 診斷與課表數據輸出")
    out_ai = gr.Textbox(label="🧠 AI 體能與疲勞診斷大腦", lines=4, elem_classes="ai-box")

    # 【第三區塊：智慧配時分段訓練計畫】
    gr.Markdown("#### 📋 智慧配時分段訓練計畫")
    with gr.Row():
        out_warmup = gr.Textbox(label="🔥 1. 暖身階段 (Warm-up)", lines=5, elem_classes="workout-card")
        out_main = gr.Textbox(label="⚡ 2. 主訓練階段 (Main Set)", lines=5, elem_classes="workout-card")
        out_cooldown = gr.Textbox(label="🧊 3. 緩和階段 (Cool Down)", lines=5, elem_classes="workout-card")

    # 【第四區塊：強度配速視覺化圖表】
    gr.Markdown("#### 📈 指標與強度分佈趨勢圖")
    out_chart = gr.Plot(label="訓練強度分佈圖")

    # ==================== 【事件綁定與初始化】 ====================

    btn_submit.click(
        fn=generate_training_schedule,
        inputs=[inp_sport, inp_run_type, inp_fatigue, inp_time],
        outputs=[out_ai, out_warmup, out_main, out_cooldown, out_chart]
    )

    # 網頁載入時自動生成初始課表
    demo.load(
        fn=generate_training_schedule,
        inputs=[inp_sport, inp_run_type, inp_fatigue, inp_time],
        outputs=[out_ai, out_warmup, out_main, out_cooldown, out_chart]
    )

# 啟動介面
demo.launch(share=True)
```
